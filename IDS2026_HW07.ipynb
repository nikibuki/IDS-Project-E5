The second step of CRISP-DM is data understanding. The goal is to collect, describe, explore, and verify the quality of the data to ensure it’s suitable for the analytical or modeling tasks ahead.

**a. Describe the data. Number of rows and columns, what does one row represent.**

```python
import pandas as pd

file1 = "9564b3c9-8da0-40bc-a15a-a845d231819c-Tootuna-arvele-votmised-kuu-jooksul-toetuste-ja-hyvitistega-kaetuse-ja-maakonna-ja-soo-ning.csv"
file2 = "601b9b09-976a-4032-a08a-add0f1268ba9-Tootud-kuu-lopu-seisuga-toetuste-ja-hyvitistega-kaetuse-ja-maakonna-ja-soo-ja-vanusegrupi-n.csv"

new_reg = pd.read_csv(file1) # new registered unemployed during the month
total_reg = pd.read_csv(file2) # registered unemployed at month end

new_reg["Kuupäev"] = pd.to_datetime(new_reg["Kuupäev"], format="%m/%d/%y")
total_reg["Kuupäev"] = pd.to_datetime(total_reg["Kuupäev"], format="%m/%d/%y")

for name, df in [("new_reg", new_reg), ("total_reg", total_reg)]:
    print(name)
    print(df.columns)
    print("Rows, columns:", df.shape)
```

**Answer:** In the first file each row is the number of newly registered unemployed people in that month, for one county, gender, age group and benefit type.
For example if we look at row 4 we find out that in 01.10.2024, in Harju county there were 105 males aged 16-24 who were newly registered for "töötutoetus".
This file has 6 columns and 85956 rows.

In the second file each row is the total number of registered unemployed people at the end of a month for one county, gender, age group, duration of their registration and benefit type.
This file has 7 columns and 213165 rows.

**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
import matplotlib.pyplot as plt

# Trend over time
new_reg.groupby("Kuupäev")["Uusi registreeritud töötuid"].sum().plot(figsize=(12, 4), title="New registrations per month")
plt.show()
total_reg.groupby("Kuupäev")["Registreeritud töötuid"].sum().plot(figsize=(12, 4), title="Registered unemployed at month end")
plt.show()

# By county (total)
total_reg.groupby("Maakond")["Registreeritud töötuid"].sum().sort_values().plot(kind="barh", figsize=(8, 6), title="Total by county")
plt.show()

# By gender, age group, duration, benefit type
fig, ax = plt.subplots(2, 2, figsize=(12, 8))
total_reg.groupby("Sugu")["Registreeritud töötuid"].sum().plot(kind="bar", ax=ax[0, 0], title="Sex")
total_reg.groupby("Vanusegrupp")["Registreeritud töötuid"].sum().plot(kind="bar", ax=ax[0, 1], title="Age group")
total_reg.groupby("Arveloleku kestus")["Registreeritud töötuid"].sum().plot(kind="bar", ax=ax[1, 0], title="Duration")
total_reg.groupby("Toetuse või hüvitise liik")["Registreeritud töötuid"].sum().plot(kind="barh", ax=ax[1, 1], title="Benefit type")
plt.tight_layout()
plt.show()

# Box plot by age group
total_reg.boxplot(column="Registreeritud töötuid", by="Vanusegrupp", figsize=(6, 4))
plt.show()
```

**Answer:** Plots are mostly unsurprising, however it's interesting that Ida-Virumaa has much more unemployed people than Tartu county despite being smaller by population. Also the economic crisis of the late 2000s is clearly visible.

**d. Check for data quality**

Missing values, duplicates, impossible or inconsistent values. Also watch out for outliers and anomalies that may require further investigation. Explain what you found and what you are going to do with the information.


```python
for name, df in [("new_reg", new_reg), ("total_reg", total_reg)]:
    print("=====", name)
    print("Missing values:\n", df.isna().sum(), "\n")
    print("Duplicate rows:", df.duplicated().sum())

# duplicates on the key columns (should be unique per combination)
key1 = ["Kuupäev", "Maakond", "Sugu", "Vanusegrupp", "Toetuse või hüvitise liik"]
key2 = key1 + ["Arveloleku kestus"]
print("Duplicate keys file 1:", new_reg.duplicated(subset=key1).sum())
print("Duplicate keys file 2:", total_reg.duplicated(subset=key2).sum())

# impossible values
print("Negative values file 1:", (new_reg["Uusi registreeritud töötuid"] < 0).sum())
print("Negative values file 2:", (total_reg["Registreeritud töötuid"] < 0).sum())

# zeros
print("Share of zeros file 1:", (new_reg["Uusi registreeritud töötuid"] == 0).mean())
print("Share of zeros file 2:", (total_reg["Registreeritud töötuid"] == 0).mean())

# outliers (simple IQR rule)
for df, col in [(new_reg, "Uusi registreeritud töötuid"), (total_reg, "Registreeritud töötuid")]:
    q1, q3 = df[col].quantile([0.25, 0.75])
    upper = q3 + 1.5 * (q3 - q1)
    print(col, "- values above", upper, ":", (df[col] > upper).sum(), "max:", df[col].max())

# unexpected category spellings (e.g. whitespace, typos)
for col in ["Maakond", "Sugu", "Vanusegrupp", "Toetuse või hüvitise liik"]:
    print(total_reg[col].value_counts(dropna=False), "\n")

# are all months present?
print(new_reg["Kuupäev"].nunique(), total_reg["Kuupäev"].nunique())
```

**Answer:** The data looks completely fine (no impossible or missing values, no duplicates etc.), which means we can use it as is, with no need to clean it.
